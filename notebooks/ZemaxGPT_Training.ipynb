```python
from unsloth import FastLanguageModel
from datasets import load_dataset
from trl import SFTTrainer, SFTConfig

DATASET_PATH = "zemax_opticstudio_finetune_dataset.jsonl"
OUTPUT_DIR = "/content/drive/MyDrive/ZemaxGPT_Llama3.2_3B"

# 1. Load Model (Unsloth handles quantization automatically)
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Llama-3.2-3B-Instruct",
    max_seq_length=1024,
    load_in_4bit=True
)

# 2. Add LoRA Adapters
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    lora_alpha=32,
    target_modules=["q_proj","k_proj","v_proj","o_proj","gate_proj","up_proj","down_proj"],
    lora_dropout=0.05,
    bias="none"
)

# 3. Load and Format Dataset
dataset = load_dataset("json", data_files=DATASET_PATH)

def format_chat(example):
    return {"text": (
        f"<|begin_of_text|><|start_header_id|>system<|end_header_id|>\n\n{example['system']}<|eot_id|>"
        f"<|start_header_id|>user<|end_header_id|>\n\n{example['instruction']}<|eot_id|>"
        f"<|start_header_id|>assistant<|end_header_id|>\n\n{example['response']}<|eot_id|>"
    )}

formatted = dataset["train"].map(format_chat)

# 4. Training Config
sft_config = SFTConfig(
    output_dir=OUTPUT_DIR,
    num_train_epochs=3,
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    fp16=True,
    logging_steps=10,
    save_strategy="steps",
    save_steps=100,
    save_total_limit=2,
    lr_scheduler_type="cosine",
    optim="adamw_8bit",
    dataset_text_field="text",
    max_seq_length=1024,
    report_to="none"
)

# 5. Train
trainer = SFTTrainer(
    model=model,
    train_dataset=formatted,
    tokenizer=tokenizer,
    args=sft_config
)

print("ðŸš€ Training starting...")
trainer.train()

# 6. Save to Google Drive
model.save_pretrained(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print("âœ… Done! Model saved to Google Drive.")
```

```python
# Cell 4: Compare Fine-Tuned vs Original (Run right after training)
from unsloth import FastLanguageModel

# â•”â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•—
# â•‘   PUT YOUR QUESTION HERE (change the text inside quotes)    â•‘
# â•šâ•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•

MY_QUESTION = "Create a tolerancing script for multistage alignment process, first focus the system, then adjust lens 2 decenter for best focus and then refocus the image plane, also provide necessary merit function if needed."

# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Load Original Model
print("Loading Original Llama 3.2 3B...")
base_model, base_tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Llama-3.2-3B-Instruct",
    max_seq_length=1024,
    load_in_4bit=True
)
FastLanguageModel.for_inference(base_model)

# Fine-tuned model is already loaded from training (model + tokenizer)
FastLanguageModel.for_inference(model)

# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
system = (
    "You are ZemaxGPT, an expert AI assistant for Zemax OpticStudio optical design. "
    "You provide precise, technically accurate guidance on sequential and non-sequential "
    "design, analysis, optimization, tolerancing, and ZOS-API programming."
)
messages = [
    {"role": "system", "content": system},
    {"role": "user", "content": MY_QUESTION}
]

# Original Model Answer
base_inputs = base_tokenizer.apply_chat_template(
    messages, tokenize=True, add_generation_prompt=True, return_tensors="pt"
).to("cuda")
base_out = base_model.generate(input_ids=base_inputs, max_new_tokens=1024, temperature=0.3, do_sample=True)
base_answer = base_tokenizer.decode(base_out[0][base_inputs.shape[-1]:], skip_special_tokens=True)

# Fine-Tuned Model Answer
ft_inputs = tokenizer.apply_chat_template(
    messages, tokenize=True, add_generation_prompt=True, return_tensors="pt"
).to("cuda")
ft_out = model.generate(input_ids=ft_inputs, max_new_tokens=1024, temperature=0.3, do_sample=True)
ft_answer = tokenizer.decode(ft_out[0][ft_inputs.shape[-1]:], skip_special_tokens=True)

# Print Both
print(f"\n{'='*70}")
print(f"QUESTION: {MY_QUESTION}")
print(f"{'='*70}")
print(f"\nðŸ”´ ORIGINAL LLAMA (Not Fine-Tuned):\n{'-'*40}")
print(base_answer)
print(f"\nðŸŸ¢ FINE-TUNED ZEMAXGPT:\n{'-'*40}")
print(ft_answer)
print(f"\n{'='*70}")
```
